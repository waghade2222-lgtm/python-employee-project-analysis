```python
# Import pandas library
import pandas as pd
# Import numpy library
import numpy as np
```

```python
# Create Project DataFrame
project = pd.DataFrame({
          "ID":["A001","A002","A003","A004","A005",
                "A002","A005","A003","A001","A003",
                "A001","A004","A004","A005"],
        "Project":["Project 1","Project 2","Project 3","Project 4","Project 5",
                   "Project 6","Project 7","Project 8","Project 9","Project 10",
                   "Project 11","Project 12","Project 13","Project 14"],
        "Cost":[1002000,2000000,4500000,5500000,np.nan,
                680000,400000,350000,np.nan,300000,
                2000000,1000000,3000000,200000],
        "Status":["Finished","Ongoing","Finished","Ongoing","Finished",
                  "Failed","Finished","Failed","Ongoing","Finished",
                  "Failed","Ongoing","Finished","Finished"]
})
# Display the Project DataFrame
print(project)
```

```python
#Creating Employee DataFrame
employee = pd.DataFrame({
           "ID":["A001","A002","A003","A004","A005"],
           "Name":["John Alter","Alice Luxemburg","Tom Sabestine","Nina Adgra","Amy Johny"],
            "Gender":["M","F","M","F","F"],
            "City":["Paris","London","Berlin","Newyork","Madrid"],
            "Age":[25,27,29,31,30]
})
#Display Employee DataFrame
print(employee)
```

```python
# Creating Seniority Level DataFrame 
seniority = pd.DataFrame({
            "ID":["A001","A002","A003","A004","A005"],
            "Designation Level": [2,2,3,2,3]
})
#Display Seniority Level DataFrame
print(seniority)
```

```python
#Saving Project DataFrame as CSV 
project.to_csv("Project.csv",index=False)
#Saving Employee DataFrame as CSV
employee.to_csv("Employee.csv",index=False)
#Saving Seniority Level DataFrame as CSV
seniority.to_csv("Seniority.csv",index=False)
print("CSV files saved successfully.")
```

```python
# Read the CSV files
project = pd.read_csv("Project.csv")
employee=pd.read_csv("Employee.csv")
seniority=pd.read_csv("Seniority.csv")
print(project)
print(employee)
print(seniority)
```

```python
#-----------------------------------------------------------
# Task 2
# Replace missing values in the Cost column
# using the running average value and for loop.
#-----------------------------------------------------------
# Traverse each row of the Project DataFrame
for i in range(len(project)):
    #Check whether the Cost value is missing (NaN)
    if pd.isna(project.loc[i,"Cost"]):
        #Calculate the running average of all previous
        #non-missing Cost values
        running_average=project.loc[:i-1,"Cost"].mean()
        #replace the missing value with the running average
        project.loc[i,"Cost"] = running_average
#Display the updated Project DataFrame 
print(project)
```

```python
#---------------------------------------------------------
#Task 3
# Split the Name column into First Name and Last Name
#and remove the original Name column
#---------------------------------------------------------
#Split the Name column into First Name and Last Name
employee[["First Name","Last Name"]]=employee["Name"].str.split(" ",n=1,expand=True)
#Remove the original Name column
employee.drop("Name",axis=1,inplace=True)
#Display the updated Employee DataFrame
print(employee)
```

```python
#----------------------------------------------------------
#Task 4
#Join all three DataFrames into single DataFrame
#and store the result in a new DataFrame named "Final"
#----------------------------------------------------------
#Merge Project and Employee DataFrames using the ID column
Final = pd.merge(project,employee,on="ID",how="left")
#Merge the above result with Seniority Level DataFrame using the ID Column
Final=pd.merge(Final,seniority,on="ID",how="left")
#Display the Final DataFrame
print(Final)
```

```python
#-------------------------------------------------------
#Task 5
#Add a Bonus column in the Final DataFrame.
#Give a 5% bonus on project Cost only to employees
#who have finished the project
#-------------------------------------------------------
#Create a new Bonus column with initial value 0
Final ["Bonus"] = 0
#Traverse each row of the Final DataFrame 
for i in range (len(Final)):
    #Check if the project status is Finished
    if Final.loc[i,"Status"]=="Finished":
        #Calculate 5% bonus on the Project Cost 
        Final.loc[i,"Bonus"]=Final.loc[i,"Cost"]*0.05
#Display the updated Final DataFrame
print(Final)
```

```python
#-----------------------------------------------------
#Task 6(Part A)
#Demote the Designation Level by 1
#for employees whose Project Status is "Failed"
#-----------------------------------------------------
#Traverse each row of the Final DataFrame
for i in range (len(Final)):
    #check if the Project Status is Failed
    if Final.loc[i,"Status"]=="Failed":
        #Demote the Designation Level by 1
        Final.loc[i,"Designation Level"]=Final.loc[i,"Designation Level"]-1
#Display the updated Final DataFrame
print(Final)

        
```

```python
#------------------------------------------------------
#Task 6 (Part B)
#Delete employee records whose
#Designation Level is above 4
#------------------------------------------------------
#keep only employees whose Designation Level is less than or equal to 4
Final = Final[Final["Designation Level"]<=4] 
#Reset the index
Final.reset_index(drop=True,inplace=True)
#Display the updated Final DataFrame 
print(Final)
```

```python
#----------------------------------------------------
#Task 7
#Add Mr. and Mrs. to the First Name column
# and drop the gender column.
#----------------------------------------------------
#Traverse each row of the Final DataFrame 
for i in range(len(Final)):
    #check if the Gender is Male
    if Final.loc[i,"Gender"]=="M":
        #Add Mr. before the First Name 
        Final.loc[i,"First Name"]="Mr."+Final.loc[i,"First Name"]
        #check if the Gender is Female
    elif Final.loc[i,"Gender"]=="F":
        #Add Mrs.before the First Name
        Final.loc[i,"First Name"]="Mrs."+Final.loc[i,"First Name"]
#Drop the Gender Column
Final.drop("Gender",axis=1,inplace=True)
#Display the updated Final DataFrame
print(Final)
        
```

```python
#------------------------------------------------------
#Task 8
# Promote Designation Level by 1
# for employees whose Age is more than 29 years.
#------------------------------------------------------
#Traverse each row of the Final DataFrame 
for i in range (len(Final)):
    #Check if the employee's Age is greater than 29
    if Final.loc[i,"Age"]>29:
        #Promote the Designation Level by 1
        Final.loc[i,"Designation Level"]=Final.loc[i,"Designation Level"]+1
#Display the updated Final DataFrame
print(Final)
        
```

```python
#---------------------------------------------------
#Task 9
#calculate the total project cost for each employee
#and store it in a new DataFrame named total_proj_cost.
#---------------------------------------------------
#Group the Final DataFrame by ID and First Name
#and calculate the total cost of all projects 
total_proj_cost=Final.groupby(["ID","First Name"])["Cost"].sum().reset_index()
#Rename the Cost column to Total Cost
total_proj_cost.rename(columns={"Cost":"Total Cost"},inplace=True)
#Display the new DataFrame
print(total_proj_cost)
```

```python
#----------------------------------------------------------
#Task 10
#Display all employee details whose City
#contains the letter "o".
#----------------------------------------------------------
#Filter the records where the city contains the letter "o"
City_o = Final[Final["City"].str.contains("o",case=False)]
#Display the Filtered employee details
print(City_o)
```
